```python
import pandas as pd
```

```python
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df['Purchase Amount (USD)']=df['Purchase Amount (USD)'].fillna(df['Purchase Amount (USD)'].mean())
```

```python
df['Review Rating']=df['Review Rating'].fillna(df['Review Rating'].mean())
```

```python
df['Previous Purchases']=df['Previous Purchases'].fillna(df['Previous Purchases'].mean())
```

```python
df['Size']=df['Size'].fillna(df['Size'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create a column age_group
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'], q=4, labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
#create column purchase_frequency_days

frequency_mapping={
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annualy':365,
    'Every 3 Months':90
}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```
